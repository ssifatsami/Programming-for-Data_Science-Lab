## Exercise 1: Course Marks with a List


```python
#storing marks
marks = [75, 82, 68, 91, 45]

#calculate total and average marks
total = 0

for mark in marks:
    total = total + mark

average = total / len(marks)

#find highest and lowest marks
highest = max(marks)
lowest = min(marks)

#count passed courses
passed = 0

for mark in marks:
    if mark >= 50:
        passed = passed + 1

#display the results
print("total Marks:", total)
print("avg Mark:", average)
print("highest Mark:", highest)
print("lowest Mark:", lowest)
print("courses Passed:", passed)
```

# Exercise 2: Weekend Temperature with a Tuple


```python

# Store temperatures for Saturday and Sunday
temperatures = (28, 30, 27, 29, 32, 26)

# Display each temperature
for temperature in temperatures:
    print("Temperature:", temperature)

# Find highest and lowest temperatures
highest = max(temperatures)
lowest = min(temperatures)

print("Highest Temperature:", highest)
print("Lowest Temperature:", lowest)
```

# Exercise 3: Word Analyzer

```python
# Take input from the user
text = input("Enter a word or short sentence: ")

# Initialize counters
vowels = 0
spaces = 0

# Examine each character
for character in text:
    if character.lower() in "aeiou":
        vowels = vowels + 1

    if character == " ":
        spaces = spaces + 1

# Display the results
print("Number of Characters:", len(text))
print("Number of Vowels:", vowels)
print("Number of Spaces:", spaces)
```

# Exercise 4: Grade Calculator Function


```python

# Define the function
def calculate_grade(marks):

    # Calculate the average
    average = sum(marks) / len(marks)

    # Determine the grade
    if average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"


# Call the function using five marks
student_marks = [85, 78, 90, 72, 80]

grade = calculate_grade(student_marks)

# Display the result
print("Marks:", student_marks)
print("Average:", sum(student_marks) / len(student_marks))
print("Grade:", grade)
```

# Exercise 5: Shopping Cart Processor


```python
# Store product details in a list of dictionaries
products = [
    {"name": "Notebook", "price": 120, "quantity": 3},
    {"name": "Pen", "price": 20, "quantity": 5},
    {"name": "USB Drive", "price": 850, "quantity": 2},
    {"name": "Mouse", "price": 650, "quantity": 1}
]

# Calculate the total cost
total = 0

for product in products:
    cost = product["price"] * product["quantity"]
    total = total + cost

# Define the discount function
def apply_discount(total):
    if total >= 2000:
        return total * 0.90
    else:
        return total

# Calculate the final price
final_price = apply_discount(total)

# Display the results
print("Total Cost:", total)
print("Final Price After Discount:", final_price)
```

# Exercise 6: Countdown

```python
# Take a number from the user
number = int(input("Enter a number: "))

# Print the countdown
while number >= 0:
    print(number)
    number = number - 1

print("Countdown finished!")
```
