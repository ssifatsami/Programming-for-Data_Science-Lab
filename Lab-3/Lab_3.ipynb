````python

# Import pandas library
import pandas as pd

# Load the Titanic dataset
df = pd.read_csv("train.csv")

print(df)
````

## **Problem 1 – Dataset Inspection**

````python
# 1.First 10 rows
print("First 10 Rows:")
print(df.head(10))

# 2.Number of rows and columns
print("\nDataset Shape:")
print(df.shape)

# 3.Column names
print("\nColumn Names:")
print(df.columns.tolist())

# 4. Data types
print("\nData Types:")
print(df.dtypes)

# 5. Descriptive statistics
print("\nDescriptive Statistics:")
print(df.describe())
````

# Problem 2 – Data **Filtering**

````python
#1.First Class Passengers
first_class = df[df["Pclass"] == 1]
print("Total First Class Passengers:", len(first_class))

#2.Passengers Younger Than 18
young_passengers = df[df["Age"] < 18]
print("Passengers Younger Than 18:")
print(young_passengers)

#3.Male Passengers Who Survived
male_survivors = df[
    (df["Sex"] == "male") &
    (df["Survived"] == 1)
]
print("Male Passengers Who Survived:")
print(male_survivors)

#4.Passengers Who Paid More Than 100
high_fare_passengers = df[df["Fare"] > 100]
print("Passengers Who Paid More Than 100:")
print(high_fare_passengers)
````

# Problem 3 – Sorting and Aggregation

````python
#1.Display 10 passengers who paid the highest fares
print("Q1: Top 10 Highest Fare Passengers")
highest_fare = df.sort_values(
    by="Fare",
    ascending=False
)
print(highest_fare.head(10))


#2.Calculate average passenger age
print("\nQ2: Average Passenger Age")
average_age = df["Age"].mean()
print("Average Age:", round(average_age, 2))


#3.Count male and female passengers
print("\nQ3: Male and Female Passenger Count")
gender_count = df["Sex"].value_counts()
print(gender_count)


#4.Calculate average fare for each passenger class
print("\nQ4: Average Fare by Passenger Class")
average_fare = df.groupby("Pclass")["Fare"].mean()
print(average_fare)


#5.Calculate survival rate for each passenger class
print("\nQ5: Survival Rate by Passenger Class")
survival_rate = df.groupby("Pclass")["Survived"].mean() * 100
print(survival_rate.round(2))
````
